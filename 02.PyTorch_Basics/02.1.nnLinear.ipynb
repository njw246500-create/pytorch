#### ▶ 인공신경망 
* nn모듈 : 신경망을 구축하고 학습시키는데 필요한 모든 구성요소를 제공하는 핵심 모듈
* 주요 역할 및 구성 요소
    1. 인공 신경망 레이어(Layer) : 입력데이터를 변환하는 다양한 층을 제공
        * 선형 레이어(nn.Linear) : 완전연결층
        * 합성곱(nn.Conv2d) : 이미지 처리에 쓰이는 CNN레이어
        * 순환 레이어(nn.LTSM,nn.GRU) : 텍스트, 시계열 처리에 쓰는 RNN 레이어
    2. 활성화 함수(Activation Functions) : 비선형성을 추가하여 모델이 복잡한 패턴을 학습 할 수 있도록 해준다. 
        * nn.ReLU(), nn.Sigmoid(), nn.Softmax() 등 
    3. 손실 함수(Loss Functions) : 모델의 예측값과 실제 값의 오차를 계산
        * nn.MESLoss(), nn.CrossEntropyLoss() 등
    4. 컨테이너 : 여러 레이어를 묶어서 하나의 모델로 구조화 
        * nn.Module : 모든 신경망 모델의 기본 클래스
        * 자신이 만드는 모든 커스텀 모델은 nn.Module을 상속받아 정의하여 사용
        * nn.Sequential : 레이어를 순차적으로 연결해 주는 간단한 래퍼임.

```python
import torch
from torch import nn
from torchviz import make_dot
```

```python
x = torch.tensor([1.])

model = nn.Linear(1,1)
print(model)
print()

print('w = ', model.weight)
print('b = ', model.bias)
print()
```

```python
fc1 = nn.Linear(1,3) # 입력 1개 출력 3개 
fc2 = nn.Linear(3,1) # 입력 3개 출력 1개

print('fc1의 w =', fc1.weight)
print('fc1의 가중치 행렬 =', fc1.weight.shape) ## 3행 1열
# w.shape = (out_features, in_features) ## 특징 

print('fc1의 b = ', fc1.bias)
print('fc1의 bias size = ', fc1.bias.shape)
print('-'*30)

print('fc2의 w = ',fc2.weight)
print('fc2의 b = ', fc2.bias) ## 바이어스는 출력갑과, 가중치는 입력값과 갯수가 같다. 
print('-'*30)

x = torch.tensor([1.]) ## 트랜스포즈 해서 3.1 행렬로 만들어서 가중합을 구하기
x = fc1(x)
print(x)
y = fc2(x)
print(y)
print('-'*30)
x = torch.tensor([1.]) # 트랜스포즈 후 바이어스 결합
y2 = ( x @ fc1.weight.T + fc1.bias ) @ fc2.weight.T + fc2.bias
print(y2)

make_dot(y)
```

Feature과 Channel의 차이
| 구분 | Feature | Channel | 
| --- | --- | --- |
| 의미 | 데이터가 가진 특성 또는 변수 | 데이터의 독립적인 데이터 경로 또는 구성요소|
| 예시 | 나이, 키, 몸무게 | RGB 이미지의 R, G, B |
| 주로 사용하는 곳 | 일반적인 데이터, 머신러닝, 완전연결층 | 이미지처리, CNN, 신호처리 | 


```python
model = nn.Linear(2,3)
# 2개의 채널(or Feature)를 가진 데이터에 대해 3개의 채널을 가지는 데이터로 변환하는 것 
x = torch.randn(2)
print(x)
print(model(x))

make_dot(model(x))
```

```python
model = nn.Linear(2,3)

x = torch.randn(5,2) # 2개의 특징(몸무게, 키)을 가지는 데이터(사람) 5개(명)   ->> 5,2 + 2.3 -> 5.3  

print(x)
model_x = model(x)
print(model_x.shape)

x = torch.randn(4,5,2)  ## linear의 시작갑과 x 의 끝값은 맞아야한다.     ->> 4,5,2 + 2,3 => 4,5,3
print(model(x).shape)     ## 2,4,5,2 + 2,3 = 2.4.5.3   -> 4,5,2 형태의 상자가 2개 -> 4,5,3 상자 2개로 표현 

x = torch.randn(2,3,6,5,3,2)
print(model(x).shape)
```

```python
fc1 = nn.Linear(1,3)
fc2 = nn.Linear(3,1)

x = torch.tensor([1.])
x = fc1(x)
y = fc2(x)

model = nn.Sequential(fc1, fc2)
x = torch.tensor([1.])
print(model(x))
```

```python
model = nn.Sequential(nn.Linear(2,5),
                      nn.Linear(5,10),
                      nn.Linear(10,3))

x = torch.randn(5, 2)
print(model(x))
```

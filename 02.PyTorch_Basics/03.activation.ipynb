#### Activation Function 
: 신경망의 선형 연산 결과에 비선형성을 추가하여 복잡한 패턴을 학습할 수 있도록 돕는 함수 

- 활성화 함수의 목적 
    - 은닉층에 비선형성 추가
    - 출력값을 조절 
    - 문제에 맞는 출력 생성

#### 1. Sigmoid 함수 
: 입력값 0~1사이의 값으로 변환하는 함수 

- 특징 
    - 출력 범위가 0과 1 사이 
    - 입력값이 0이면 출력값은 0.5
    - 입력값이 커지면 출력은 1에 가까워짐 
    - 입력값이 작아지면 출력은 0에 가까워짐 
    - 이진분류에서 한 클래스에 속할 확률처럼 해헉할 수 있는 값들을 만들 때 사용 
    
- 단점 
    - 입력값의 절댓값이 아주 커지면 기울기 작아지는 포화현상이 나타남. 
    - 은닉층에 여러번 사용하면 기울기 소실 문제가 생길 수 있음

```python
import torch
import torch.nn as nn 
```

```python
x = torch.tensor([-10., -2, -1, 0, 1, 2, 10]) ## 왼쪽은 0에 가까움, 오른쪽은 1에 가까움 

output = torch.sigmoid(x)
print(output)
```

```python
logits = torch.tensor([-2.0, -0.5, 0, 0.8, 2])
print('모델을 통과한 출력(Logits) : ', logits)

torch.sigmoid(logits)
```

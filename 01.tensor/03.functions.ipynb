#### ▶ PyTorch 주요 수학 함수

#### 1. 기본 수학 함수
| 함수 | 의미 | 예시 | 설명 
|:---:|:---:|:---:|:---|
| torch.abs(a)| 절댓값 | abs(-3) -> 3 | 모든 원소의 절댓값 반환 |
| torch.sqrt(a) | 제곱근 | sqrt(9) -> 3 | $\sqrt{x}$ |
| torch.square(a) | 제곱 | square(3) -> 9 | $x^2$ |
| torch.pow(a,n) | 거듭제곱 | pow(2,3) -> 8 | $x^n$ |
| torch.exp(a) | 지수함수 | exp(1) -> e | $e^x$ | 
| torch.log(a) | 자연로그 | log(e) -> 1 | ln(x) | 
| torch.log10(a) | 상용로그 | log10(10) -> 1 | $\log_{10}(x)$ | 
| torch.log2(a) | 밑이 2인 로그 | log2(2) -> 1 | $\log_2(x)$ | 

```python
a = torch.randn(3,3)
print(a)

print('절댓값 : \n', torch.abs(a) )
print('제곱근 : \n', torch.sqrt(a))
print('제곱 : \n', torch.square(a))
print('거듭제곱 : \n', torch.pow(a,3))

# 지수함수
print('지수함수 : \n',torch.exp(a)) # 2.718^x
print('자연로그 : \n',torch.log(torch.abs(a))) # logeX 
print('자연로그 : \n',torch.log(torch.exp(torch.tensor(1))))
print('상용로그 : \n',torch.log10(torch.tensor(10)))
print('밑이 2인 로그 : \n',torch.log2(torch.tensor(2)))
```

#### 3. 최솟값 / 최댓값

| 함수 | 의미 | 설명 |  
|:---:|:---:|:---:|
| torch.min(a)| 최솟값 | 전체 원소 중 가장 작은값 |
| torch.max(a)| 최댓값 | 전체 원소 중 가장 큰값 |
| torch.argmin(a)| 최솟값의 인덱스 | 최솟값이 위치한 인덱스 반환 |
| torch.argmax(a)| 최댓값의 인덱스 | 최댓값이 위치한 인덱스 반환 | 


```python
a = torch.rand(3,4)
print(a)
```

```python
print('최댓값 :', torch.max(a))
# 옵션 dim : 각 차원에서 가장 큰 값과 그 값의 인덱스 반환
print(torch.max(a, dim = 0)) # 열끼리 
print(torch.max(a, dim = 1)) # 행끼리
print('-'*20)

# 옵션 keepdims : 최댓값을 반환한 후에도 텐서의 차원을 그대로 유지할 것인지 여부(기본값 : False)
print(torch.max(a, dim=0, keepdim=True))
print(torch.max(a, dim = 1, keepdim=True))
print('-'*50)

print(a)
print(torch.min(a))
print(torch.min(a, dim=0))
print(torch.min(a, dim=1))
print('-'*50)

print(torch.argmax(a))
```

#### 4. 최솟값 / 최댓값

| 함수 | 의미 | 설명 |  
|:---:|:---:|:---:|
| torch.mean(a)| 평균 | 모든 원소의 평균 |
| torch.sum(a)| 합계 | 모든 원소의 합 |
| torch.prod(a)| 곱 | 모든 원소의 곱 |
| torch.median(a)| 중앙값 | 정렬했을 때 가운데 값 |
| torch.std(a)| 표준편차 | 데이터의 분산 정도 |
| torch.var(a)| 분산 | 데이터의 분산 |

```python
print(a)

print('합계 : ', torch.sum(a))
print(torch.sum(a, dim=1))
print(torch.sum(a, dim=1, keepdim=True))
print('-'*50)

print('평균 : ', torch.mean(a))
print(torch.mean(a, dim=1))
print(torch.mean(a, dim=1, keepdim=True))
print('-'*50)

# 데이터가 짝수이면 더 작은값을 선택
print('중앙값 : ',torch.median(a))
print('표준편차 : ',torch.std(a))
print('분산 : ',torch.var(a))
```

#### ▶ torch.nan
- not a number(숫자가 아님) : 수학적으로 정의할 수 없거나 컴퓨터가 정상적인 실수값으로 표현할 수 없는 연산 불능 / 수치적인 오류 상태를 의미 
    - 정의되지 않은 수학
        - 연산음수의 로그 계산 : torch.log(torch.tensor(-1.0)) -> nan
        - 0을 0으로 나누기 : torch.tensor(0) / torch.tensor(0) -> nan
        - 무한대 간의 뺄셈 : inf - inf -> nan 

    - 딥러닝 학습 중 수치적 불안정성(Gradient Explosion 등) 
        - 그라디언트 폭주 : 가중치 업데이트량이 너무 커져 수치가 표현 범위 초과할 때 
        - 언더플로우 / 오버플로우 

```python
torch.nan

print(torch.log(torch.tensor(-1)))

print(torch.isnan(torch.tensor([1, 2, torch.nan, 3, 4])))
print(torch.isinf(torch.tensor([1,2,3,4, torch.inf])))
```

#### ▶ reshape()
: 텐서의 원소 개수는 유지한 채 모양(shape)만 변경할 때 사용

```python
a = torch.randint(1,5,size=(12,))
print(a)
print(a.shape)

b = a.reshape(2,2,3)
print(b)
```

```python
a = torch.arange(20, )
print(a)
print(a.reshape(4,5))
print('-'*50)
print(a.reshape(4,-1)) ## 열은 너가 알아서 맞춰줘 = -1
print(a.reshape(2,-1))
print(a.reshape(2,-1).shape)

print(a.reshape(2,5,-1))
print(a.reshape(2,-1,5))
print(a.reshape(2,-1,5).shape)
```

#### ▶ Numpy배열을 tensor로 변환 
: 임베딩 벡터 처리, 데이터 전처리, 모델 입력 데이터 구성 등에서 많이 사용

```python
import numpy as np
import torch
```

- torch.from_numpy(array)
  : 메모리 공유, numpy 배열을 tensor로 변환

```python
arr = np.array([1.0,2,3,4])

x = torch.from_numpy(arr)

print('numpy 배열 :', arr)
print('tensor :', x)

arr[0] = 100
print('변경후')
print('numpy 배열 :', arr)
print('tensor :', x)
```

- torch.tensor(array)
    : 데이터를 복사, 원본과 분리된 tensor 생성

```python
arr = np.array([1.0,2,3,4])

x = torch.tensor(arr)

print('변경전')
print('numpy 배열 :', arr)
print('tensor :', x)

arr[0] = 100

print('변경후')
print('numpy 배열 :', arr)
print('tensor :', x)
```

- torch.as_tensor(array)
    : 가능하면 메모리를 공유하고, 필요한 경우에만 데이터를 복사하는 함수 
    - torch.from_numpy()와 비슷하게 동작하지만, 입력 자료형이나, 장치(device)등을 지정할 수 있다는 차이

```python
x = torch.tensor([1.0,2,3],device = 'cuda')

# GPU -> CPU -> NUMPY 
arr = x.cpu().numpy()
```

#### 문제 02

실험 장비에서 측정한 양수 데이터가 다음과 같음.

```python
measurement = torch.tensor([1.0, 4.0, 9.0, 16.0])
```

**요구사항**
1. 각 원소의 제곱근을 `sqrt_data`에 저장.
2. 각 원소의 제곱을 `square_data`에 저장.

#### 문제 03

제품 무게를 측정한 결과가 다음과 같음. (단위: kg)

```python

```

**요구사항**
1. `torch.abs()`를 이용해 절댓값 계산.
2. `torch.round()`를 이용해 정수 단위로 반올림.
3. `torch.round()`의 `decimals` 옵션을 이용해 소수점 둘째 자리까지 반올림.

#### 문제 04

세 지점에서 측정한 네 종류의 센서값이 다음과 같음.

```python
sensor = torch.tensor([[12, 25, 18, 31],
                       [ 9, 28, 22, 17],
                       [15, 19, 35, 24]])
```

**요구사항**
1. 전체 데이터의 최솟값과 최댓값 출력.
2. `dim=0`을 기준으로 각 열의 최댓값과 해당 인덱스 출력.
3. `dim=1`을 기준으로 각 행의 최솟값과 해당 인덱스 출력.
4. 전체 텐서에서 최댓값과 최솟값의 위치 인덱스를 각각 출력.

````python
sensor = torch.tensor([[12, 25, 18, 31],
                       [ 9, 28, 22, 17],
                       [15, 19, 35, 24]])
print('최댓값 : ', torch.max(sensor))
print('최솟값 : ', torch.min(sensor))
print('dim = 0 : ', torch.max(sensor, dim = 0))
print('dim = 1 : ', torch.min(sensor, dim = 1))
print('최댓값의 위치 : ', torch.argmax(sensor))
print('최솟값의 위치 : ', torch.argmin(sensor))
````

#### 문제 05

세 팀의 일주일 생산량을 나타내는 텐서가 다음과 같음.

```python
production = torch.tensor([[12., 15., 18., 10.],
                           [20., 17., 16., 19.],
                           [ 8., 11., 13., 14.]])
```

**요구사항**
1. 전체 생산량의 합계와 평균 계산.
2. 팀별 생산량 합계를 `dim=1` 기준으로 계산.
3. 팀별 평균 생산량을 `dim=1` 기준으로 계산하고 `keepdims=True` 적용.
4. 전체 생산량의 곱, 중앙값, 표준편차, 분산 계산.
5. 각 결과를 출력.

````python
production = torch.tensor([[12., 15., 18., 10.],
                           [20., 17., 16., 19.],
                           [ 8., 11., 13., 14.]])
print("전체 생산량의 합계 : ", torch.sum(production))
print("전체 생산량의 평균 : ", torch.mean(production))
print("팀별 생산량 기준 함계 dim = 1 :" , torch.sum(production, dim=1))
print("팀별 평균생산량 :", torch.mean(production, dim=1, keepdim=True))
print("전체 생산량의 곱 :", torch.prod(production))
print("전체 생산량의 중앙값 : ", torch.median(production))
print("전체 생산량의 표준편차 : ", torch.std(production))
print("전체 생산량의 분산 : ", torch.var(production))
````
